## Experiment 1 – Import, Load, and View a Dataset

**Objective:** Write a program to import necessary libraries, load a standard dataset, and view its contents including shape, columns, data types, and sample records.

### Step 1: Import Required Libraries

```python
# -----------------------------------------------------------
# EXPERIMENT 1 - Step 1: Import Required Libraries
# -----------------------------------------------------------

# pandas  : data manipulation and analysis (DataFrames)
import pandas as pd

# numpy   : numerical computing (arrays, math operations)
import numpy as np

# seaborn : statistical data visualisation; also provides sample datasets
import seaborn as sns

# matplotlib : base plotting library
import matplotlib.pyplot as plt

# sklearn datasets : provides classic ML benchmark datasets
from sklearn.datasets import load_iris

print("All libraries imported successfully!")
print(f"   pandas  version : {pd.__version__}")
print(f"   numpy   version : {np.__version__}")
print(f"   seaborn version : {sns.__version__}")
```

### Step 2: Load the Dataset

```python
# -----------------------------------------------------------
# EXPERIMENT 1 - Step 2: Load the Iris Dataset
# -----------------------------------------------------------

# Load the Iris dataset from sklearn
# The dataset contains measurements of 150 iris flowers across 3 species
iris_bunch = load_iris()

# Convert the sklearn Bunch object to a pandas DataFrame for easier analysis
# feature_names gives us the column labels
df = pd.DataFrame(
    data=iris_bunch.data,
    columns=iris_bunch.feature_names
)

# Add the target (species) column and map numeric codes to actual flower names
df['species'] = iris_bunch.target            # 0, 1, 2  (numeric labels)
df['species'] = df['species'].map(
    {0: 'setosa', 1: 'versicolor', 2: 'virginica'}
)

print("Dataset loaded successfully!")
print(f"   Dataset name : Iris Dataset")
print(f"   Source       : sklearn.datasets.load_iris()")
```

### Step 3: View the Dataset

```python
# -----------------------------------------------------------
# EXPERIMENT 1 - Step 3a: Dataset Shape and Column Names
# -----------------------------------------------------------

print("Dataset Shape (rows, columns):", df.shape)
print()
print("Column Names:")
for i, col in enumerate(df.columns, start=1):
    print(f"   {i}. {col}")
```

```python
# -----------------------------------------------------------
# EXPERIMENT 1 - Step 3b: View the First 10 Rows (Head)
# df.head(n) returns the first n rows of the DataFrame
# -----------------------------------------------------------

print("First 10 rows of the dataset:")
df.head(10)
```

```python
# -----------------------------------------------------------
# EXPERIMENT 1 - Step 3c: View the Last 10 Rows (Tail)
# df.tail(n) returns the last n rows of the DataFrame
# -----------------------------------------------------------

print("Last 10 rows of the dataset:")
df.tail(10)
```

```python
# -----------------------------------------------------------
# EXPERIMENT 1 - Step 3d: View Random Sample Rows
# df.sample(n) returns n randomly selected rows
# random_state ensures reproducibility of the random selection
# -----------------------------------------------------------

print("Random sample of 5 rows:")
df.sample(5, random_state=42)
```

```python
# -----------------------------------------------------------
# EXPERIMENT 1 - Step 3e: View Data Types of Each Column
# dtypes tells us whether each column is int, float, object (string), etc.
# -----------------------------------------------------------

print("Data Types of each column:")
print(df.dtypes)
```

```python
# -----------------------------------------------------------
# EXPERIMENT 1 - Step 3f: Visualise the Dataset (Pair Plot)
# A pair plot shows scatter plots for every pair of numeric features.
# Diagonal cells show the KDE (distribution) of each feature.
# Colour-coding by 'species' helps see if classes are separable.
# -----------------------------------------------------------

print("Generating Pair Plot...")
sns.pairplot(
    df,
    hue='species',           # colour each species differently
    diag_kind='kde',         # kernel density estimate on diagonal
    plot_kws={'alpha': 0.7}  # slight transparency to reduce overplotting
)
plt.suptitle("Iris Dataset - Pair Plot", y=1.02, fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
print("Pair plot displayed.")
```

### Experiment 1 – Conclusion

- The **Iris dataset** was successfully imported using `sklearn.datasets` and converted into a pandas `DataFrame`.
- The dataset contains **150 rows** and **5 columns** (4 numeric feature columns + 1 categorical target column).
- We viewed the data using `head()`, `tail()`, `sample()`, and inspected data types via `dtypes`.
- The pair plot confirms that the **setosa** species is linearly separable from the other two.

## Experiment 2 – Display the Summary and Statistics of the Dataset

**Objective:** Compute and display descriptive statistics, check for missing values, examine class distribution, compute a correlation matrix, and visualise feature distributions.

### Step 1: Dataset Info (Structure Overview)

```python
# -----------------------------------------------------------
# EXPERIMENT 2 - Step 1: df.info()
# info() provides a concise summary of the DataFrame:
#   - Number of rows and columns
#   - Column names, non-null counts, and data types
#   - Memory usage
# -----------------------------------------------------------

print("Dataset Info:")
print("-" * 50)
df.info()
```

### Step 2: Descriptive Statistics

```python
# -----------------------------------------------------------
# EXPERIMENT 2 - Step 2a: Descriptive Statistics for Numeric Columns
# describe() computes: count, mean, std deviation, min,
# 25th/50th/75th percentiles, and max for every numeric column.
# -----------------------------------------------------------

print("Descriptive Statistics (Numeric Features):")
df.describe()
```

```python
# -----------------------------------------------------------
# EXPERIMENT 2 - Step 2b: Descriptive Statistics for ALL Columns
# include='all' also shows statistics for categorical columns:
#   - unique : number of distinct values
#   - top    : most frequent value
#   - freq   : frequency of the most common value
# -----------------------------------------------------------

print("Descriptive Statistics (All Columns):")
df.describe(include='all')
```

### Step 3: Check for Missing Values

```python
# -----------------------------------------------------------
# EXPERIMENT 2 - Step 3: Missing Value Analysis
# isnull()  - boolean mask (True where value is NaN/None)
# sum()     - count of missing values per column
# Knowing missing values is critical before training any ML model.
# -----------------------------------------------------------

missing_values = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    'Missing Count'  : missing_values,
    'Missing %'      : missing_percent.round(2)
})

print("Missing Value Report:")
print(missing_report)
print()
if missing_values.sum() == 0:
    print("No missing values found. The dataset is complete!")
else:
    print(f"Total missing values: {missing_values.sum()}")
```

### Step 4: Class Distribution (Target Variable)

```python
# -----------------------------------------------------------
# EXPERIMENT 2 - Step 4: Class Distribution
# value_counts() counts how many rows belong to each category of 'species'.
# A balanced dataset (equal samples per class) is ideal for classification.
# -----------------------------------------------------------

class_dist = df['species'].value_counts()
class_percent = df['species'].value_counts(normalize=True) * 100

class_report = pd.DataFrame({
    'Count'      : class_dist,
    'Percentage' : class_percent.round(2)
})

print("Class Distribution of 'species':")
print(class_report)
print()
print("The dataset is perfectly balanced (50 samples per class = 33.33% each).")
```

### Step 5: Correlation Matrix

```python
# -----------------------------------------------------------
# EXPERIMENT 2 - Step 5: Correlation Matrix
# corr() computes the Pearson correlation coefficient between every
# pair of numeric features. Values range from -1 (perfect negative)
# to +1 (perfect positive correlation). Values near 0 = no relationship.
# -----------------------------------------------------------

# Select only numeric columns for correlation
numeric_df = df.select_dtypes(include=[np.number])
corr_matrix = numeric_df.corr()

print("Pearson Correlation Matrix:")
print(corr_matrix.round(3))
print()

# Visualise the correlation matrix as a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,          # show correlation values in each cell
    fmt='.2f',           # format to 2 decimal places
    cmap='coolwarm',     # blue = negative, red = positive correlation
    linewidths=0.5,
    square=True
)
plt.title("Iris Dataset - Correlation Heatmap", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### Step 6: Feature Distribution Plots

```python
# -----------------------------------------------------------
# EXPERIMENT 2 - Step 6a: Histogram of All Numeric Features
# Histograms reveal the distribution shape of each feature:
#   Normal (bell-shaped), skewed, bimodal, uniform, etc.
# The dashed black line marks the mean of each feature.
# -----------------------------------------------------------

numeric_features = numeric_df.columns.tolist()

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle("Iris Dataset - Feature Distributions (Histograms)",
             fontsize=14, fontweight='bold')

colors = ['#4C72B0', '#DD8452', '#55A868', '#C44E52']

for ax, feature, color in zip(axes.flatten(), numeric_features, colors):
    ax.hist(df[feature], bins=20, color=color, edgecolor='white', alpha=0.85)
    ax.set_title(feature, fontsize=11)
    ax.set_xlabel("Value (cm)")
    ax.set_ylabel("Frequency")
    mean_val = df[feature].mean()
    ax.axvline(mean_val, color='black', linestyle='--', linewidth=1.5,
               label=f'Mean = {mean_val:.2f}')
    ax.legend(fontsize=9)

plt.tight_layout()
plt.show()
```

```python
# -----------------------------------------------------------
# EXPERIMENT 2 - Step 6b: Box Plots (per Species)
# Box plots show: median, IQR (25th-75th percentile), whiskers,
# and outliers. Plotting per species helps compare how features
# differ across classes.
# -----------------------------------------------------------

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle("Iris Dataset - Box Plots by Species",
             fontsize=14, fontweight='bold')

for ax, feature in zip(axes.flatten(), numeric_features):
    sns.boxplot(
        data=df,
        x='species',
        y=feature,
        palette='Set2',
        ax=ax
    )
    ax.set_title(feature, fontsize=11)
    ax.set_xlabel("Species")
    ax.set_ylabel("Value (cm)")

plt.tight_layout()
plt.show()
```

### Step 7: Grouped Statistics per Species and Skewness/Kurtosis

```python
# -----------------------------------------------------------
# EXPERIMENT 2 - Step 7a: Grouped Statistics per Species
# groupby('species') splits the DataFrame into groups.
# agg() then applies multiple aggregation functions to each
# numeric feature, giving a detailed per-class statistical profile.
# -----------------------------------------------------------

print("Grouped Statistics by Species (Mean | Std | Min | Max):")
print("=" * 70)

grouped_stats = df.groupby('species')[numeric_features].agg(
    ['mean', 'std', 'min', 'max']
).round(3)

grouped_stats
```

```python
# -----------------------------------------------------------
# EXPERIMENT 2 - Step 7b: Skewness and Kurtosis
#
# Skewness : measures asymmetry of the distribution
#   0       = perfectly symmetric
#   Positive = right-skewed (long tail on right)
#   Negative = left-skewed  (long tail on left)
#
# Kurtosis : measures the 'tailedness' of the distribution
#   0       = normal distribution (mesokurtic)
#   Positive = heavier tails / more outliers (leptokurtic)
#   Negative = lighter tails (platykurtic)
# -----------------------------------------------------------

skewness = numeric_df.skew().round(4)
kurtosis = numeric_df.kurt().round(4)

shape_report = pd.DataFrame({
    'Skewness' : skewness,
    'Kurtosis' : kurtosis
})

print("Skewness and Kurtosis of Each Feature:")
print(shape_report)
```

### Experiment 2 – Conclusion

The following statistics were computed and displayed for the **Iris Dataset**:

| Statistic | Method Used | Observation |
|---|---|---|
| Structure Overview | `df.info()` | 150 rows, 5 columns, no null values |
| Descriptive Stats | `df.describe()` | Mean petal length = 3.76 cm, high variance |
| Missing Values | `df.isnull().sum()` | 0 missing values — dataset is clean |
| Class Distribution | `value_counts()` | Perfectly balanced (50 samples/class) |
| Correlation | `df.corr()` + heatmap | Petal length and petal width are highly correlated (r = 0.96) |

**Key Insight:** The high correlation between petal length and petal width, and the clear separability of *Iris setosa* from the other species, suggests that a simple classifier (e.g., Decision Tree or Logistic Regression) can achieve high accuracy on this dataset.
